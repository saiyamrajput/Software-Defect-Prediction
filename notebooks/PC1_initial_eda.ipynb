## Day 1

```python
import pandas as pd
from scipy.io import arff

# Load the arff file
data = arff.loadarff('../data/raw/pc1.arff')

# Convert to a pandas DataFrame
df = pd.DataFrame(data[0])

print(df.head())
```

```python
print("Shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())
```

```python
df["defects"].value_counts()
```

```python
df["defects"].value_counts(normalize=True) * 100
```

```python
df.describe()
```

## Initial EDA Findings
* The class distribution of the data set is imbalanced.
* The PC1 dataset consists of 1109 rows and 22 columns. Rows contain information on software modules and columns contain attributes of teh respective software modules.
* There are 22 attribute columns out of which 21 columns are used as explanatory variates (input of the model) and 1 is response variate (output of the model).
* There are no missing values found in the dataset.
* There are total of 1032 non-defective modules and 77 defective modules out of 1109 software modules.
* Several explanatory variables are rightly skewed because their means are larger than their medians.

## Day 2

```python
X = df.drop(columns=["defects"])
Y = df["defects"]

print(X.shape)
print(Y.shape)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42, stratify=Y)
```

```python
print(X_train.shape)
print(X_test.shape)
print(Y_train.shape)
print(Y_test.shape)
```

```python
Y.value_counts(normalize=True)
Y_train.value_counts(normalize=True)
Y_test.value_counts(normalize=True)
```

## Day 3

```python
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, precision_score, recall_score, f1_score, roc_auc_score
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
```

```python
Y_train = Y_train.astype(str)
Y_test = Y_test.astype(str)

dummy = DummyClassifier(strategy="most_frequent")
```

```python
dummy.fit(X_train, Y_train)
dummy_prediction = dummy.predict(X_test)

print(dummy_prediction[:20])
```

```python
dummyAccuracy = accuracy_score(Y_test, dummy_prediction)
dummyConfusionMatrix = confusion_matrix(Y_test, dummy_prediction)

print("Accuracy:", dummyAccuracy)
print("Confusion Matrix:")
print(dummyConfusionMatrix)
```

```python
precision = precision_score(Y_test, dummy_prediction, pos_label="true", zero_division=0)
recall = recall_score(Y_test, dummy_prediction, pos_label="true")
f1 = f1_score(Y_test, dummy_prediction, pos_label="true", zero_division=0)

print("Precision:", precision)
print("Recall:", recall)
print("F1-score:", f1)
```

```python
pipe = make_pipeline(StandardScaler(), LogisticRegression())
```

```python
pipe.fit(X_train, Y_train)
pipe_prediction = pipe.predict(X_test)
pipeProbability = pipe.predict_proba(X_test)
pipe_defectiveProbability = pipeProbability[:, 1]

print(pipe_prediction[:20])

print("\nClasses:")
print(pipe.classes_)

print("\nProbability shape:")
print(pipeProbability.shape)

print("\nFirst 5 probability rows:")
print(pipeProbability[:5])
```

```python
pipeAccuracyScore = accuracy_score(Y_test, pipe_prediction)
pipeConfusionMatrix = confusion_matrix(Y_test, pipe_prediction)
pipePrecision = precision_score(Y_test, pipe_prediction, pos_label="true", zero_division=0)
pipeRecall = recall_score(Y_test, pipe_prediction, pos_label="true")
pipeF1 = f1_score(Y_test, pipe_prediction, pos_label="true", zero_division=0)
pipeROC_AUC = roc_auc_score(Y_test, pipe_defectiveProbability)

print("Accuracy: ", pipeAccuracyScore)
print("\nConfusion Matrix: ")
print(pipeConfusionMatrix)

print("\nPrecision:", pipePrecision)
print("\nRecall:", pipeRecall)
print("\nF1-score:", pipeF1)
print("\nROC-AUC Score:", pipeROC_AUC)
```

## Day 3 Findings

The DummyClassifier baseline achieved about 93.2% accuracy but detected 0 defective modules. Logistic Regression achieved about 92.8% accuracy and detected 1 defective modules, but it still missed 14 of the 15 defective modules in the test set. This shows that Logistic Regression does not outperform the baseline in terms of accuracy, but it does improve defect detection by identifying one defective module that the baseline missed. Therefore, accuracy alone does not give a complete picture of the model's performance, especially because the dataset is imbalanced.

## Day 4

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
randomForest = RandomForestClassifier(random_state=42)
```

```python
randomForest.fit(X_train, Y_train)

randomForest_prediction = randomForest.predict(X_test)
randomForest_probability = randomForest.predict_proba(X_test)
randomForest_defectiveProbability =randomForest_probability[:, 1]

print(randomForest_prediction[:20])

print("\nClasses:")
print(randomForest.classes_)

print("\nProbability shape:")
print(randomForest_probability.shape)

print("\nFirst 5 probability rows:")
print(randomForest_probability[:5])
```

```python
randomForest_confusionMatrix = confusion_matrix(Y_test, randomForest_prediction)
randomForest_accuracy = accuracy_score(Y_test, randomForest_prediction)
randomForest_precision = precision_score(Y_test, randomForest_prediction, pos_label="true", zero_division=0)
randomForest_recall = recall_score(Y_test, randomForest_prediction, pos_label="true", zero_division=0)
randomForest_F1 = f1_score(Y_test, randomForest_prediction, pos_label="true", zero_division=0)
randomForest_ROC_AUC = roc_auc_score(Y_test, randomForest_defectiveProbability)

print("Confusion Matrix: ")
print(randomForest_confusionMatrix)
print("\nAccuracy: ", randomForest_accuracy)
print("\nPrecision:", randomForest_precision)
print("\nRecall:", randomForest_recall)
print("\nF1-score:", randomForest_F1)
print("\nROC-AUC Score:", randomForest_ROC_AUC)
```

## Day 4 Findings

Random Forest achieved higher accuracy (94.1%) than Logistic Regression (92.8%). Random Forest detected 5 of the 15 defective modules, while Logistic Regression detected only 1 of 15. This means Random Forest achieved a higher recall of 33.3%, compared with 6.7% for Logistic Regression.

Random Forest had higher precision, F1-score, and ROC-AUC than Logistic Regression. Therefore, Random Forest performed better than Logistic Regression on PC1 under this train/test split.

However, Random Forest still missed 10 of the 15 defective modules. This means that although Random Forest performed better than Logistic Regression, it still did not detect a large proportion of the actual defects. This shows accuracy alone is not sufficient for evaluating software defect prediction models.

## Day 5

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.model_selection import cross_validate
from sklearn.metrics import make_scorer, precision_score, recall_score, f1_score
```

```python
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
```

```python
metric_names = {
    "accuracy": "accuracy",
    "precision": make_scorer(precision_score, pos_label="true"),
    "recall": make_scorer(recall_score, pos_label="true"),
    "f1": make_scorer(f1_score, pos_label="true"),
    "roc_auc": "roc_auc"
}
```

```python
baseline_cv = cross_validate(estimator=dummy, X=X_train, y=Y_train, cv=skf, scoring=metric_names)
print(baseline_cv, "\n")
```

```python
print("Accuracy mean:", baseline_cv["test_accuracy"].mean())
print("Accuracy std:", baseline_cv["test_accuracy"].std())

print("\nPrecision mean:", baseline_cv["test_precision"].mean())
print("Precision std:", baseline_cv["test_precision"].std())

print("\nRecall mean:", baseline_cv["test_recall"].mean())
print("Recall std:", baseline_cv["test_recall"].std())

print("\nF1 mean:", baseline_cv["test_f1"].mean())
print("F1 std:", baseline_cv["test_f1"].std())

print("\nROC-AUC mean:", baseline_cv["test_roc_auc"].mean())
print("ROC-AUC std:", baseline_cv["test_roc_auc"].std())
```

```python
logistic_cv = cross_validate(estimator=pipe, X=X_train, y=Y_train, cv=skf, scoring=metric_names)
print(logistic_cv, "\n")
```

```python
print("Accuracy mean:", logistic_cv["test_accuracy"].mean())
print("Accuracy std:", logistic_cv["test_accuracy"].std())

print("\nPrecision mean:", logistic_cv["test_precision"].mean())
print("Precision std:", logistic_cv["test_precision"].std())

print("\nRecall mean:", logistic_cv["test_recall"].mean())
print("Recall std:", logistic_cv["test_recall"].std())

print("\nF1 mean:", logistic_cv["test_f1"].mean())
print("F1 std:", logistic_cv["test_f1"].std())

print("\nROC-AUC mean:", logistic_cv["test_roc_auc"].mean())
print("ROC-AUC std:", logistic_cv["test_roc_auc"].std())
```

```python
randomForest_cv = cross_validate(estimator=randomForest, X=X_train, y=Y_train, cv=skf, scoring=metric_names)
print(randomForest_cv, "\n")
```

```python
print("Accuracy mean:", randomForest_cv["test_accuracy"].mean())
print("Accuracy std:", randomForest_cv["test_accuracy"].std())

print("\nPrecision mean:", randomForest_cv["test_precision"].mean())
print("Precision std:", randomForest_cv["test_precision"].std())

print("\nRecall mean:", randomForest_cv["test_recall"].mean())
print("Recall std:", randomForest_cv["test_recall"].std())

print("\nF1 mean:", randomForest_cv["test_f1"].mean())
print("F1 std:", randomForest_cv["test_f1"].std())

print("\nROC-AUC mean:", randomForest_cv["test_roc_auc"].mean())
print("ROC-AUC std:", randomForest_cv["test_roc_auc"].std())
```

```python
models = ["DummyClassifier", "Logistic Regression", "Random Forest Classifier"]
meanTableAccuracy = [baseline_cv["test_accuracy"].mean(), logistic_cv["test_accuracy"].mean(), randomForest_cv["test_accuracy"].mean()]
meanTablePrecision = [baseline_cv["test_precision"].mean(), logistic_cv["test_precision"].mean(), randomForest_cv["test_precision"].mean()]
meanTableRecall = [baseline_cv["test_recall"].mean(), logistic_cv["test_recall"].mean(), randomForest_cv["test_recall"].mean()]
meanTableF1 = [baseline_cv["test_f1"].mean(), logistic_cv["test_f1"].mean(), randomForest_cv["test_f1"].mean()]
meanTableRocAuc = [baseline_cv["test_roc_auc"].mean(), logistic_cv["test_roc_auc"].mean(), randomForest_cv["test_roc_auc"].mean()]

meanComparisonTable = pd.DataFrame({
    "Model": models,
    "Accuracy Mean": meanTableAccuracy,
    "Precision Mean": meanTablePrecision,
    "Recall Mean": meanTableRecall,
    "F1 Mean": meanTableF1,
    "ROC-AUC Mean": meanTableRocAuc
})

print(meanComparisonTable)
```

```python
stdTableAccuracy = [baseline_cv["test_accuracy"].std(), logistic_cv["test_accuracy"].std(), randomForest_cv["test_accuracy"].std()]
stdTablePrecision = [baseline_cv["test_precision"].std(), logistic_cv["test_precision"].std(), randomForest_cv["test_precision"].std()]
stdTableRecall = [baseline_cv["test_recall"].std(), logistic_cv["test_recall"].std(), randomForest_cv["test_recall"].std()]
stdTableF1 = [baseline_cv["test_f1"].std(), logistic_cv["test_f1"].std(), randomForest_cv["test_f1"].std()]
stdTableRocAuc = [baseline_cv["test_roc_auc"].std(), logistic_cv["test_roc_auc"].std(), randomForest_cv["test_roc_auc"].std()]

stdComparisonTable = pd.DataFrame({
    "Model STD": models,
    "Accuracy STD": stdTableAccuracy,
    "Precision STD": stdTablePrecision,
    "Recall STD": stdTableRecall,
    "F1 STD": stdTableF1,
    "ROC-AUC STD": stdTableRocAuc
})

print(stdComparisonTable)
```

```python
finalComparisonTable = pd.DataFrame({
    "Model": models,
    "Accuracy": [f"{meanTableAccuracy[i]:.3f} ± {stdTableAccuracy[i]:.3f}" for i in range(3)],
    "Precision": [f"{meanTablePrecision[i]:.3f} ± {stdTablePrecision[i]:.3f}" for i in range(3)],
    "Recall": [f"{meanTableRecall[i]:.3f} ± {stdTableRecall[i]:.3f}" for i in range(3) ],
    "F1": [f"{meanTableF1[i]:.3f} ± {stdTableF1[i]:.3f}" for i in range(3)],
    "ROC-AUC": [f"{meanTableRocAuc[i]:.3f} ± {stdTableRocAuc[i]:.3f}" for i in range(3)]
})

print(finalComparisonTable)
```

## Day 5 Findings

The DummyClassifier achieved about 93.0% accuracy, but it had zero precision, recall, and F1 with a ROC-AUC of 0.5. This shows that the high accuracy comes from predicting the majority class while failing to detect defective modules. Both Logistic Regression and Random Forest performed better than the DummyClassifier for defect detection.
Random Forest had higher precision (0.530), recall (0.238), F1-score (0.327), and slightly higher ROC-AUC (0.843) than Logistic Regression. Therefore, Random Forest was stronger overall for defect detection under the 5-fold cross-validation evaluation. However, its mean recall of about 24% shows that it still missed many defective modules. This again shows why accuracy alone is not sufficient for evaluating software defect prediction models on an imbalanced dataset.


## Day 6

```python
randomForest_class_weight = RandomForestClassifier(random_state=42, class_weight="balanced")
```

```python
randomForest_class_weight.fit(X_train, Y_train)

randomForest_prediction_class_weight = randomForest_class_weight.predict(X_test)
randomForest_probability_class_weight = randomForest_class_weight.predict_proba(X_test)
randomForest_defectiveProbability_class_weight =randomForest_probability_class_weight[:, 1]

print(randomForest_prediction_class_weight[:20])

print("\nClasses:")
print(randomForest_class_weight.classes_)

print("\nProbability shape:")
print(randomForest_probability_class_weight.shape)

print("\nFirst 5 probability rows:")
print(randomForest_probability_class_weight[:5])
```

```python
randomForest_confusionMatrix_class_weight = confusion_matrix(Y_test, randomForest_prediction_class_weight)
randomForest_accuracy_class_weight = accuracy_score(Y_test, randomForest_prediction_class_weight)
randomForest_precision_class_weight = precision_score(Y_test, randomForest_prediction_class_weight, pos_label="true", zero_division=0)
randomForest_recall_class_weight = recall_score(Y_test, randomForest_prediction_class_weight, pos_label="true", zero_division=0)
randomForest_F1_class_weight = f1_score(Y_test, randomForest_prediction_class_weight, pos_label="true", zero_division=0)
randomForest_ROC_AUC_class_weight = roc_auc_score(Y_test, randomForest_defectiveProbability_class_weight)

print("Confusion Matrix: ")
print(randomForest_confusionMatrix_class_weight)
print("\nAccuracy: ", randomForest_accuracy_class_weight)
print("\nPrecision:", randomForest_precision_class_weight)
print("\nRecall:", randomForest_recall_class_weight)
print("\nF1-score:", randomForest_F1_class_weight)
print("\nROC-AUC Score:", randomForest_ROC_AUC_class_weight)
```

## Day 6 Findings:

The balanced Random Forest detected 8 of the 15 defective modules, compared with 5 of 15 for the original Random Forest. Class weighting increased recall from 0.333 to 0.533 and reduced false negatives from 10 to 7, showing that the model became better at detecting defective modules. However, precision decreased slightly from 0.625 to 0.615 because false positives increased from 3 to 5. The F1-score improved from 0.435 to 0.571, while accuracy increased slightly from 0.941 to 0.946 and ROC-AUC increased from 0.875 to 0.896. Therefore, class weighting improved defect detection by increasing recall and F1-score, while also slightly improving accuracy and ROC-AUC.
