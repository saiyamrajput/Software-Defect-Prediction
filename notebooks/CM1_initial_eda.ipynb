```python
import pandas as pd
from scipy.io import arff

# Load the arff file
data = arff.loadarff('../data/raw/cm1.arff')

# Convert to a pandas DataFrame
df = pd.DataFrame(data[0])

print(df.head())
```

```python
print("Shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())
```

## Initial EDA Findings
* The class distribution of the data set is imbalanced.
* The CM1 dataset consists of 498 rows and 22 columns. Rows contain information on software modules and columns contain attributes of teh respective software modules.
* There are 22 attribute columns out of which 21 columns are used as explanatory variates (input of the model) and 1 is response variate (output of the model).
* There are no missing values found in the dataset.
* There are total of 449 non-defective modules and 49 defective modules out of 498 software modules.
* The data is positively rightly skewed.
